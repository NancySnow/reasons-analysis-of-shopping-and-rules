# 规则组 check

核对旧 ENV 的资格底稿，以及本轮 `rules-groups.csv` 的完整名单和新增 ENV。

```python
import pandas as pd

X_DATA = "../数据/02_X匹配数据"
environment_metadata = pd.read_csv(f"{X_DATA}/environment_metadata.csv")
rules_groups = pd.read_csv(f"{X_DATA}/rules-groups.csv")

old_envs = set(environment_metadata["ENV"])
new_envs = rules_groups.loc[~rules_groups["ENV"].isin(old_envs)].copy()

print(f"旧 ENV 底稿：{len(environment_metadata)} 组")
print(f"本轮规则组总名单：{len(rules_groups)} 组")
print(f"本轮新增 ENV：{len(new_envs)} 组")

print("\n旧 ENV 资格底稿：")
display(environment_metadata)
print("\n本轮新增 ENV：")
display(new_envs if not new_envs.empty else pd.DataFrame({"提示": ["没有新增 ENV"]}))
```

```python
table_pivot = rules_groups.pivot_table(
    index="event_year",
    columns=["trading_platform", "event"],
    values="ENV",
    aggfunc="count",
    fill_value=0,
)

table_pivot
```
