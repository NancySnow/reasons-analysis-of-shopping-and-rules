# 规则组覆盖检查

读取 `environment_metadata.csv`，按纵轴年份、横轴交易平台与活动生成记录数交叉表。结果仅在 Notebook 中显示，不导出文件。

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# 支持从 lasso论文/ 或 lasso论文/python/ 启动 Jupyter
candidates = [Path.cwd(), Path.cwd().parent]
project_root = next(
    (path for path in candidates if (path / "数据" / "02_X匹配数据" / "environment_metadata.csv").is_file()),
    None,
)
if project_root is None:
    raise FileNotFoundError(
        "找不到 environment_metadata.csv；请从 lasso论文 或 lasso论文/python 目录启动 Jupyter。"
    )

metadata_path = project_root / "数据" / "02_X匹配数据" / "environment_metadata.csv"
metadata = pd.read_csv(metadata_path, encoding="utf-8-sig")
print(f"读取文件：{metadata_path}")
print(f"记录数：{len(metadata)}")
```

```python
required_columns = ["event_year", "trading_platform", "event"]
missing_columns = [column for column in required_columns if column not in metadata.columns]
if missing_columns:
    raise KeyError(f"environment_metadata.csv 缺少字段：{missing_columns}")

check_data = metadata[required_columns].copy()
if check_data.isna().any().any():
    missing_counts = check_data.isna().sum()
    raise ValueError(f"交叉表字段存在缺失值：\n{missing_counts[missing_counts > 0]}")

check_data["event_year"] = pd.to_numeric(check_data["event_year"], errors="raise").astype(int)
check_data = check_data.sort_values(required_columns).reset_index(drop=True)
display(check_data.rename(columns={
    "event_year": "年份",
    "trading_platform": "交易平台",
    "event": "活动",
}))
```

```python
cross_table = pd.crosstab(
    index=check_data["event_year"],
    columns=[check_data["trading_platform"], check_data["event"]],
    dropna=False,
).sort_index()
cross_table.index.name = "年份"
cross_table.columns.names = ["交易平台", "活动"]

display(
    cross_table.style
    .set_caption("environment_metadata 规则组数量交叉表")
    .format(precision=0)
    .set_properties(**{"text-align": "center"})
)
```
