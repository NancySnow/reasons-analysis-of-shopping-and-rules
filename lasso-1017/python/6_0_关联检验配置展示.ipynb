# 6_0 关联检验配置展示

把关联检验配置的每个末端字段完整展开为 DataFrame。

```python
from pathlib import Path
import json
import pandas as pd

PYTHON_DIR = Path.cwd() if Path.cwd().name == 'python' else Path.cwd() / 'python'
config = json.loads((PYTHON_DIR / '关联检验配置.json').read_text(encoding='utf-8'))

def flatten(value, prefix=''):
    if isinstance(value, dict):
        for key, child in value.items():
            yield from flatten(child, f'{prefix}.{key}' if prefix else key)
    else:
        yield prefix, value

config_df = pd.DataFrame(flatten(config), columns=['配置项', '具体值'])
display(config_df)
```
