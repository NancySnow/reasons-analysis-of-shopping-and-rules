# 5_0 弱标签词典展示

把弱标签词典的全部内容拆成表格查看。

```python
from pathlib import Path
import json
import pandas as pd

PYTHON_DIR = Path.cwd() if Path.cwd().name == 'python' else Path.cwd() / 'python'
dictionary = json.loads((PYTHON_DIR / '弱标签词典.json').read_text(encoding='utf-8'))
```

```python
anchors_df = pd.DataFrame(dictionary['anchors'].items(), columns=['锚点类型', '正则表达式'])
display(anchors_df)

components_df = pd.DataFrame([{
    '组件编号': item['component_id'],
    '定义': item['persisted_definition']['定义'],
    '纳入条件': item['include'],
    '排除条件': item['exclude'],
    '解释': item['interpretation'],
} for item in dictionary['components']])
display(components_df)

for item in dictionary['components']:
    for rule in item['candidate_rules']:
        print(f"{item['component_id']}：{rule['regex']}")
        print()
```

```python
outcomes_df = pd.DataFrame(dictionary['outcomes']).merge(
    pd.DataFrame(dictionary['display_names'].items(), columns=['outcome_id', '显示名称']),
    on='outcome_id', how='left'
)
display(outcomes_df)

other_sections = {key: value for key, value in dictionary.items()
                  if key not in {'anchors', 'components', 'outcomes', 'display_names'}}
details_df = pd.DataFrame([(key, json.dumps(value, ensure_ascii=False, indent=2))
                           for key, value in other_sections.items()],
                          columns=['词典部分', '具体内容'])
display(details_df)
```
